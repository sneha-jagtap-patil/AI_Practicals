```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(12) #seed value is 12
x= np.linspace(0, 1, 20)
y = np.cos(x) + 0.1*np.random.rand(20) #0.1 differnce between values....downscaling..........at the end 20 or 0 ypu can add
p= np.poly1d(np.polyfit(x,y,1)) #order of polynomials
print(p)

t= np.linspace(0,1,200) #for smooth line 200
plt.plot(x,y,'bo',t,p(t),'-') #bo-blue circle.....rgbcymk...black-k..........markers----s-square....v...for trangle.....o for circle......plottingline...-,--, styles
plt.xlabel('x values')
plt.ylabel('function to be appriximated')
plt.title('curve fitting')
plt.show()
```
